# Run the CPU reservoir experiment
MVP uses rate states, not biological spike recordings. See README for limitations.

```python
from pathlib import Path
import os
if not Path('configs').exists():
    os.chdir('..')
import json
from datetime import datetime, timezone
from threadpoolctl import threadpool_limits
from flying.training.train import run
config = json.loads(Path('configs/mvp.json').read_text())
config['seeds'] = [42]
out = 'outputs/notebook_' + datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S%f')
with threadpool_limits(limits=1):
    results = run(config, out)
print(results)
```
